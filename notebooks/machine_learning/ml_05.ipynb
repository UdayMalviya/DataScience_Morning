```python
import numpy as  np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('../../data/job_salary_prediction_dataset.csv')
```

```python
df
```

```python
sns.scatterplot(df, x='experience_years', y='salary')
```

```python
sns.kdeplot(df['salary'])
```

```python
X = df.drop('salary', axis=1)
y = df['salary']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=42)
```

```python
# standardscaling 
# mean, std/
scale = StandardScaler()
scale.fit_transform(X_train[['experience_years']])
```

```python
sns.boxplot(X_train['experience_years'])
```

```python
arr = scale.fit_transform(pd.DataFrame(y_train))
```
