````python
df = pd.read_csv('../../data/CarPrice_Assignment.csv')
````

````python
df.info()
````

````python
df.nunique()
````

````python
df.select_dtypes('string')
````

````python
X =  df.drop(columns=['CarName', 'price'])
y = df['price']
````

````python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
````

````python
num = X.select_dtypes('number').columns
cat = X.select_dtypes('string').columns
````

````python
num_pipe = Pipeline([
    ('scale', StandardScaler())
])
cat_pipe = Pipeline([('encode', OneHotEncoder(sparse_output=False, drop='first', handle_unknown='ignore'))])
````

````python
transform = ColumnTransformer([
    ("n_preprocess", num_pipe, num),
    ("cat_preprocess", cat_pipe, cat)
])
````

````python
from sklearn.feature_selection import VarianceThreshold
````

````python
model = Pipeline([
    ('transform', transform),
    ('variance', VarianceThreshold(0.3)),
    ('model', LinearRegression())
])
````

````python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
````

````python
model.fit(X_train, y_train)
````

````python
y_pred = model.predict(X_test)
````

````python
r2_score(y_test, y_pred)
````
