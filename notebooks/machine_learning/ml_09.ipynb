```python
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_features=100, n_samples=100000, noise=100)
```

```python
import seaborn as sns
```

```python
X.shape, y.shape
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2)
```

```python
model = Lasso(alpha=0.01)
model2 = LinearRegression()
model3 = Ridge(alpha=0.01)
```

```python
model.fit(X_train, y_train)
model2.fit(X_train, y_train)
model3.fit(X_train, y_train)
```

```python
print(f"coef of lasso: {model.coef_}, intercept of lasso: {model.intercept_}")
```

```python
print(f"coef of linear_reg: {model2.coef_}, intercept of linear_reg: {model2.intercept_}")
```

```python
print(f"coef of ridge: {model3.coef_}, intercept of ridge: {model3.intercept_}")
```

```python
from sklearn.metrics import r2_score
```

```python
# model.predict(X_test)
print(r2_score(y_test, model.predict(X_test)))
print(r2_score(y_test, model2.predict(X_test)))
print(r2_score(y_test, model3.predict(X_test))) # ridge
```

```python
model = ElasticNet(alpha=0.08, l1_ratio=0.9)
```

```python
model.fit(X_train, y_train)
```

```python
r2_score(y_test, model.predict(X_test))
```
